```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r'C:\Users\user\Desktop\data-science-journey\jupyter notebooks files\cleaned_marketing_campaign.csv')
df.head()
```

Campaign analysis

```python
df = df[df['Conversions'] > 0]
df
```

```python
# the channel that performed best

df.groupby('Channel').agg({
    'conversion_rate': 'mean',
    'Spend': 'sum',
    'Conversions': 'sum',
    'click_through_rate': 'mean',
    'cost_acquisition': 'mean'
}).sort_values('conversion_rate', ascending = False)

# from this, it is clear that instagram performed best because 540k was spent on instagram and the conversion rate is 0.126 which has a ~0.001 difference from the 
# highest which spent almost 3x what was spent on instagram.
```

```python
# campaign tag that perform best

df.groupby('Campaign_Tag').agg({
    'Spend': 'sum',
    'Conversions': 'sum',
    'cost_acquisition': 'mean'
}). sort_values('Conversions', ascending = False)

# from this, it is clear that TI campaign tag performed best, consumed less money ~651k and conversion of ~64k that is almost 3x lower than FA which consumed 1.65M
# with a conversion of ~66k
```

```python
# more spend equals more conversions?

df[['Conversions', 'Spend']].corr()

# more spend does not equal conversion. from the analysis, to get a conversion of 1 you spend 0.057 
# and conversion of 0.057 you spend 1.
```
