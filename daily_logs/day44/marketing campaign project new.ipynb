DATA CLEANING

```python
# clean the column names of white soaces

df.rename(columns = {' Campaign_ID ': 'Campaign_ID', 'Clicks ': 'Clicks'}, inplace = True)
df.head()
```

```python
# check for duplicates

df.duplicated().sum()
```

```python
# remove duplicates

df = df.drop_duplicates()
df
```

```python
# drop the duplicated clicks column

df = df.loc[:, ~df.columns.duplicated()]
df.head()
```

```python
# cleaning each columns and changing data types

# start date

df['Start_Date'] = pd.to_datetime(df['Start_Date'], format = 'mixed', errors = 'coerce')
df.head()
```

```python
# end date

df['End_Date'] = pd.to_datetime(df['End_Date'], format = 'mixed', errors = 'coerce')
df.head()
```

```python
df['End_Date'].dtypes
```

```python
# check and clean channel column

df['Channel'].unique()
```

```python
# correct the spelling of some of the data

df['Channel'] = df['Channel'].replace({'Tik_Tok': 'TikTok', 'Facebok':  'Facebook', 'Insta_gram': 'Instagram', 'Gogle': 'Google Ads', 'E-mail': 'Email'})
df.head()
```

```python
df['Channel'].isnull().sum()
```

```python
# fill the missing value on the channel column with 'unknown'

df['Channel'] = df['Channel'].fillna('Unknown')
df['Channel'].value_counts()
```

```python
# check the impressions column


df['Impressions'].dtypes
```

```python
# check the clicks column

df['Clicks'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
# look into the spend column

df['Spend'].dtypes
```

```python
# remove the $ sign and convert to numeric

df['Spend'] = df['Spend'].str.lstrip('$')
df.head()
```

```python
df['Spend'] = pd.to_numeric(df['Spend'], errors = 'coerce')
df['Spend'].dtypes
```

```python
# look into the conversion column

df['Conversions'].isnull().sum()
```

```python
# drop the null values

df = df.dropna(subset = ['Conversions'])
df['Conversions'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
df
```

```python
# look into the campaign tags 

df['Campaign_Tag'].unique()
```

```python
# look into the Active column

df['Active'].unique()
```

```python
# replace y, yes, true with 1 and no, false with 0
# for easy calculation later on during EDA

df['Active'] = df['Active'].replace({'Y': 1, 'No': 0, 'True': 1, 'Yes': 1, 'False': 0})
df
```

```python
df['Active'] = df['Active'].astype(int)
df.head()
```

Logical validati

```python
# 

(df['Clicks'] > df['Conversions']).sum()
```

```python
# cannot have negative spend, so we remove it

(df['Spend'] < 0).sum()
```

```python
df = df[df['Spend'] >= 0]
df
```

```python
# check the end and start date

(df['End_Date'] > df['Start_Date']).sum()
```

```python
# a campaign cannot end when it has not even started

(df['Start_Date'] > df['End_Date']).sum()
```

```python
# 99 rolls where the end is highe than the start date

df[df['Start_Date'] > df['End_Date']]
```

```python
# swapped the end date with start date

mask = df['Start_Date'] >  df['End_Date']

df.loc[mask, ['Start_Date', 'End_Date']] = df.loc[mask, ['End_Date', 'Start_Date']].values
```

```python
(df['Start_Date'] > df['End_Date']).sum()
```

```python
df
```

PERFORMANCE METRICS

```python
# feature engineering

df['Duration'] = df['End_Date'] - df['Start_Date']
df.head()
```

```python
# click through rate

df['click_through_rate'] = df['Clicks']/ df['Impressions']
df.head()
```

```python
# calculate the conversion rate

df['conversion_rate'] = df['Conversions']/df['Clicks']
df.head()
```

```python
# cost per click

df['cost_per_click'] = df['Spend']/df['Clicks']
df.head()
```

```python
df['cost_acquisition'] = df['Spend']/df['Conversions']
df.head()
```
