```python
np.random.seed(0)


# Print a formatted table header
student_score = np.random.randint(30, 100, size = (30, 5))
subject = ['math', 'english', 'science', 'history', 'cs']


header = f"{'students':<10}" + "".join(f"{sub:>10}" for sub in subject)

print(header)
print('-' * 60)


# Print each student's row

for i, row in enumerate(student_score):
    score_str = "".join(f"{val:>10}" for val in row)
    print(f"{'student'+ str(i + 1):<10}" + score_str)
```

```python
# 1. CLASS SUMMARY
#    - Mean, median, std, min, max for each subject

student_score.mean(axis = 0)
```

```python
student_score.std(axis = 0)
```

```python
student_score.min(axis = 0)
```

```python
student_score.max(axis = 0)
```

```python
np.median(student_score, axis = 0)
```

```python
# 2. STUDENT RANKINGS
#    - Each student's overall average

student_avg = student_score.mean(axis = 1)

student_avg
```

```python
rank = np.argsort(student_avg)[:: -1]

rank
```

```python
#    - Print top 3 and bottom 3 students with their averages

print('top 3 best students')

for i, student_idx in enumerate(rank[:3]):
    print(f" #{i+1}: student {student_idx} - average : {student_avg[student_idx]:.1f}")
```

```python
#    - Rank students from best to worst (use argsort)
print('Bottom 3 students')

for i, student_idx in enumerate(rank[-3:]):
    print(f" last #{len(rank)-1}: student{student_idx} - average: {student_avg[student_idx]:.1f}")
```

```python
# 3. PERFORMANCE FLAGS
#    - Identify students who are "at risk": average below 50

at_risk = np.where(student_avg < 50)[0]
print('at risk students:', at_risk)
print('their averages:', student_avg[at_risk])
```

```python
#    - Identify "high performers": average above 80

high_perform = np.where(student_avg > 80)[0]
print('high performancce students:', high_perform)
print('their score:', student_avg[high_perform])
```

```python
#    - Identify students who have at least one score below 40 ("needs attention")

low_student = np.where((student_score < 40).any(axis = 1))[0]
print('need s attention:', low_student)
print('their score:', student_avg[low_student])
```

```python
# 4. NORMALIZATION
#    - Normalize each subject to 0-1 range using min-max scaling:
#      scaled = (x - min) / (max - min)

student_score.mean(axis = 0)
```

```python
mins = student_score.min(axis = 0)
mins
```

```python
maxi = student_score.max(axis = 0)

maxi
```

```python
scale = (student_score - mins) / (maxi - mins)

scale
```

```python
scale.mean(axis = 0)
```

```python
#    - Show the mean of each subject before and after normalization


print('mean before normalization')
print(dict(zip(subject, student_score.mean(axis = 0).round(2))))
```

```python
print("\nMean per subject AFTER normalization:")
print(dict(zip(subject, scale.mean(axis=0).round(2))))
```

```python
# 5. INSIGHT
#    - Which subject has the highest failure rate? (score < 50)

failure_rate = (student_score < 50).mean(axis = 0) * 100
worst_subject_idx = np.argmax(failure_rate)

print('failure rate:', failure_rate)
print(f"Highest failure rate: {subject[worst_subject_idx]} at {failure_rate[worst_subject_idx]:.1f}")
```

```python
#    - Which subject has the most consistent scores (lowest std)?

stds = student_score.std(axis = 0)
most_consistent_idx = np.argmin(stds)

print(f" most consistent subject: {subject[most_consistent_idx]} with std={stds[most_consistent_idx]:.2f}")
```

```python
high_rate = (student_score > 80).mean(axis = 0)* 100
best_subject_idx = np.argmax(high_rate)

print('high rate:', high_rate)
print(f" best score subject: {subject[best_subject_idx]} at {high_rate[best_subject_idx]:.1f}")
```
