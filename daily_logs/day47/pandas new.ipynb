```python
# 1. DATA OVERVIEW
#    Run a proper first_look() function on the dataset

def first_look(df):
    print('data shape', df.shape)
    print('\n --- First 5 rolls---' )
    print(df.head())
    print('\n --- the data types ---')
    print(df.dtypes)
    print('\n --- missing values ---')
    print(df.isnull().sum())
    print('\n --- statistics of data ---')
    print(df.describe())

first_look(df)
```

```python

# 2. ENRICHMENT
#    Add these columns:
#    - 'average' (mean of 3 subjects)

df['average'] = df[['math', 'english', 'science']].mean(axis = 1).round(2)
```

```python
df.head()
```

```python
#    - 'grade' (A/B/C/D/F using pd.cut)

df['grade'] = pd.cut(
    df['average'],
    bins = [0, 50, 60, 70, 80, 100],
    labels = ['F', 'D', 'C', 'B', 'A']
)
```

```python
df.head()
```

```python
#    - 'status' (Distinction / Pass / Fail using np.select)

condition = [
    df['average'] > 80,
    df['average'] >= 60,
    df['average'] < 60
]
status = ['Distinction', 'Pass', 'Fail']

df['status'] = np.select(condition, status, default = 'Unknown')
df.head()
```

```python
#    - 'strongest_subject' (which of math/english/science
#       is highest for each student — hint: use df[cols].idxmax(axis=1))


subject_cols = ['math', 'english', 'science']
df['strongest_subject'] = df[subject_cols].idxmax(axis = 1) 


df.head()
```

```python
#    - 'strongest_subject' (which of math/english/science
#       is highest for each student — hint: use df[cols].idxmax(axis=1))

df[['name', 'math', 'english', 'science', 'strongest_subject']].head(10)
```

```python
# 3. CLASS COMPARISON
#    For each class (A/B/C), show:
#    - Total students

df.groupby('grade')['name'].count()
```

```python
#    - Average score

df.groupby('grade').agg(
    count = ('average', 'count'),
    mean_avg = ('average', 'mean'),
    best_math = ('math', 'max'),
    best_english = ('english', 'max'),
    best_science = ('science', 'max'),
    worst_math = ('math', 'min'),
    worst_english = ('english', 'min'),
    worst_science = ('science', 'min')
)
```

```python
#    - Pass rate (%)
# Correct — count students passing divided by total students


pass_rate = (df['average'] >= 50).sum() / len(df) * 100

print(f'over all pass rate: {pass_rate:.1f}%')
```
