```python
# function that list the stats of a data, like mean, min, max


def list_stats(numbers):
    if len(numbers) == 0:
        return {
            'mean' : None,
            'Max' : None,
            'Min' : None
        }

    total = sum(numbers)
    len_counts = len(numbers)
    number_avg = total / len_counts
    Max = max(numbers)
    Min = min(numbers)

    return {
        'mean': number_avg,
        'max' : Max,
        'min' : Min
    }
```

```python
list_stats([10, 20, 30, 40])
```

```python
# function for standard deviation



def std_dev(number):
    if len(number) == 0:
        return None

    mean_value = sum(number)/ len(number)

    variance = 0
    for num in number:
        variance += (num - mean_value)**2
    variance = variance / len(number)

    std = math.sqrt(variance)
    return std
```

```python
std_dev([10, 20, 30, 40])
```

```python
# function to get the second largest number

def second_largest(numbers):
    if len(numbers) < 2:
        return None

    largest_num = numbers[0]
    second_largest = None

    for num in numbers:
        if num > largest_num:
            second_largest = largest_num
            largest_num = num

        elif num != largest_num:
              if second_largest is None or  num > second_largest:
                second_largest = num
                
    return second_largest
    
        
```

```python
# function to find the smallest number

def find_second_smallest(numbers):
    if len(numbers) < 2:
        return None

    smallest =  numbers[0]
    second = None

    for num in  numbers:
        if num < smallest:
                second = smallest
                smallest = num

        elif num != smallest:
                if second is None or num < second:
                    second = num

    return second
```

```python
# function for frequency report

def frequency_report(numbers):

    count_num = {}

    for num in numbers:
        if num in count_num:
            count_num[num] += 1
        else:
            count_num[num] = 1

    most_frequent = None
    highest_count = 0

    for num in count_num:
        if count_num[num] > highest_count:
            highest_count = count_num[num]
            most_frequent = num


    least_frequent = None
    lowest_count = float('inf')

    for num, count in count_num.items():
        if count < lowest_count:
            lowest_count = count
            least_frequent = num

    
    seen = set()
    unique_count = 0

    for num in numbers:
        if num not in seen:
            seen.add(num)
            unique_count += 1


    total_count = 0

    for num in numbers:
        total_count += 1

    return {
        'most frequent': most_frequent,
        'least frequent': least_frequent,
        'unique count': unique_count,
        'total count': total_count
    }
```

```python
frequency_report([1,2,2,3,1,1,4])
```

```python
# function for percentage of frequency

def frequency_percentage(numbers):

    count_num = {}

    for num in numbers:
        if num in count_num:
            count_num[num] += 1
        else:
            count_num[num] = 1

    total_count = len(numbers)


    percentage = {}

    for num in count_num:
        percent = (count_num[num]/ total_count) * 100
        percentage[num] = round(percent, 2)

    return percentage
```

```python
# function for click through rate

def click_through_rate(impressions, clicks):
    if impressions == 0:
        return 0

    ctr = (clicks/impressions) * 100

    return round(ctr, 2)
```

```python
# function for cost per acquisition

def cost_per_acquisition(spend, conversions):
    if conversions == 0:
        return 0

    cpa = spend/conversions

    return round(cpa, 2)
```

```python
# return on ad spend


def return_on_ad_spend(revenue, spend):
    if spend == 0:
        return None

    roas = revenue / spend

    return round(roas, 2)
```

```python
def profit(revenue, spend):
    if revenue == 0:
        return None

    profit = revenue - spend

    return profit
```

```python
def profit_margin(revenue, spend):
    if revenue == 0:
        return None

    profit_margin = ((revenue - spend) / revenue) * 100

    return round(profit_margin, 2)
```

```python
# function for campaign performance

def campaign_performance(impressions, clicks, conversions, spend, revenue):

    if impressions == 0:
        ctr = None
    else:
        ctr = round((clicks/impressions) * 100,2 )

    if clicks == 0:
        conversion_rate = None
    else:
        conversion_rate = round((conversions / clicks) * 100, 2)

    if conversions == 0:
        cpa = None
    else:
        cpa = round(spend / conversions, 2)

    if spend == 0:
         roas = None
    else:
        roas = round(revenue / spend, 2)


    profit = revenue - spend


    if revenue == 0:
         profit_margin = None
    else:
        profit_margin = round(((revenue - spend) / revenue) * 100, 2)


    return {
        'click through rate': ctr,
        'conversion rate': conversion_rate,
        'cost per acquisition': cpa,
        'return on ad spend': roas,
        'profit': profit,
        'profit margin': profit_margin,
        'revenue': revenue,
        'spend': spend
    }
```
